### Import Modules

```python
# Importing pandas and matplotlib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Read in the Netflix CSV as a DataFrame
netflix_df = pd.read_csv("netflix_data.csv")
```

### Filter the DataFrame

- Movies released in 1990's (1990 - 1999)

```python
filtered_df = netflix_df[netflix_df['type'] == 'Movie']
filtered_df = filtered_df[np.logical_and(filtered_df['release_year'] >= 1990, filtered_df['release_year'] <= 1999)]
```

### Find the most frequent duration

```python
duration_dict = {}

# Save each duration to a key in dictionary
for d in filtered_df['duration']:
    if d in duration_dict:
        duration_dict[d] += 1
    else:
        duration_dict[d] = 1

# Find the duration with the highest frequency
duration = max(duration_dict, key=duration_dict.get)
```

### Count the number of short action movies

- A movie is considered a short movie if it is less than 90 minutes

```python
EXCLUSIVE_LIMIT = 90

short_action_movies = filtered_df[
    np.logical_and(
        filtered_df["duration"] < EXCLUSIVE_LIMIT, filtered_df["genre"] == "Action"
    )
]

short_movie_count = short_action_movies.shape[0]
```
