The Nobel Prize has been among the most prestigious international awards since 1901. Each year, awards are bestowed in chemistry, literature, physics, physiology or medicine, economics, and peace. In addition to the honor, prestige, and substantial prize money, the recipient also gets a gold medal with an image of Alfred Nobel (1833 - 1896), who established the prize.

![](Nobel_Prize.png)

The Nobel Foundation has made a dataset available of all prize winners from the outset of the awards from 1901 to 2023. The dataset used in this project is from the Nobel Prize API and is available in the `nobel.csv` file in the `data` folder.

In this project, you'll get a chance to explore and answer several questions related to this prizewinning data. And we encourage you then to explore further questions that you're interested in!

```python
# Loading in required libraries
import pandas as pd
import seaborn as sns
import numpy as np

nobel_df = pd.read_csv("./data/nobel.csv")
```

## What is the most commonly awarded gender and birth country?

```python
top_gender = nobel_df["sex"].value_counts().idxmax()
top_country = nobel_df["birth_country"].value_counts().idxmax()

print(f"Top Gender: {top_gender}")
print(f"Top Country: {top_country}")
```

## Which decade had the highest ratio of US-born Nobel Prize winners to total winners in all categories?

```python
nobel_df["decade"] = (nobel_df["year"] // 10) * 10

# Create a boolean flag
nobel_df["is_usa"] = nobel_df["birth_country"] == "United States of America"

# Group by decade and calculate the mean of the boolean flag
# (The mean of a boolean series represents the ratio/percentage of True values)
max_decade_usa = nobel_df.groupby("decade")["is_usa"].mean().idxmax()

print(f"Decade with the highest ratio of US-born Nobel Prize winners: {max_decade_usa}")
```

## Who was the first woman to receive a Nobel Prize, and in what category?

```python
nobel_df["year_won"] = nobel_df["prize"].str.split(" ").str[-1]
nobel_df["year_won"] = pd.to_numeric(nobel_df["year_won"]).astype('Int64')
women_laureats = nobel_df[nobel_df["sex"] == "Female"]

first_woman_index = women_laureats["year_won"].idxmin()

first_woman_name = nobel_df["full_name"].loc[first_woman_index]
first_woman_category = nobel_df["category"].loc[first_woman_index]

print(f"The first woman to won the Nobel Prize is {first_woman_name} with a category of {first_woman_category}")
```
